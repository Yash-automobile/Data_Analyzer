```python
import numpy as np


class DataAnalytics:

    def __init__(self,data):
        self.__data = np.array(data)

    # ---------------- ARRAY CREATION ----------------

    @staticmethod
    def create_array():
        print("\nChoose Array Type")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = input("Enter choice: ")

        if choice == "1":
            values = input("Enter values separated by spaces: ")
            return np.array(list(map(float,values.split())))

        elif choice == "2":
            rows = int(input("Enter number of rows: "))
            columns = int(input("Enter number of columns: "))

            values = input("Enter values separated by spaces: ")
            values = list(map(float,values.split()))

            if len(values) != rows * columns:
                print("Number of values does not match.")
                return None

            return np.array(values).reshape(rows,columns)

        elif choice == "3":
            layers = int(input("Enter number of layers: "))
            rows = int(input("Enter number of rows: "))
            columns = int(input("Enter number of columns: "))

            values = input("Enter values separated by spaces: ")
            values = list(map(float,values.split()))

            if len(values) != layers * rows * columns:
                print("Number of values does not match.")
                return None

            return np.array(values).reshape(
                layers,
                rows,
                columns
            )

        else:
            print("Invalid choice.")
            return None

    # ---------------- INDEXING AND SLICING ----------------

    def indexing(self):
        index = int(input("Enter index: "))

        if self.__data.ndim == 1:

            if 0 <= index < len(self.__data):
                print("Value:",self.__data[index])
            else:
                print("Invalid index.")

        else:
            print("For 2D or 3D arrays, use NumPy indexing.")
            print("Array:")
            print(self.__data)

    def slicing(self):
        start = int(input("Enter start index: "))
        end = int(input("Enter end index: "))

        print("Sliced Array:")
        print(self.__data[start:end])

    # ---------------- COMBINE AND SPLIT ----------------

    def combine(self,other):

        if self.__data.ndim != other.ndim:
            print("Arrays must have the same dimensions.")
            return

        if self.__data.ndim == 1:
            print("Combined Array:")
            print(np.concatenate((self.__data,other)))

        elif self.__data.ndim == 2:

            if self.__data.shape[1] == other.shape[1]:
                print("Combined Array:")
                print(np.concatenate((self.__data,other)))
            else:
                print("Number of columns must be the same.")

        else:
            print("Combine is supported for 1D and 2D arrays.")

    def split(self):
        parts = int(input("Enter number of parts: "))

        if parts <= 0:
            print("Parts must be greater than 0.")
            return

        result = np.array_split(self.__data,parts)

        print("\nSplit Arrays:")

        for i in range(len(result)):
            print("Part",i + 1,":",result[i])

    # ---------------- MATHEMATICAL OPERATIONS ----------------

    def mathematical_operations(self):

        print("\nMathematical Operations")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        choice = input("Enter choice: ")

        if choice in ["1","2","3","4","5","6"]:

            print("\nEnter second array:")
            other = DataAnalytics.create_array()

            if other is None:
                return

        if choice == "1":

            if self.__data.shape == other.shape:
                print("Result:")
                print(self.__data + other)
            else:
                print("Arrays must have the same shape.")

        elif choice == "2":

            if self.__data.shape == other.shape:
                print("Result:")
                print(self.__data - other)
            else:
                print("Arrays must have the same shape.")

        elif choice == "3":

            if self.__data.shape == other.shape:
                print("Result:")
                print(self.__data * other)
            else:
                print("Arrays must have the same shape.")

        elif choice == "4":

            if self.__data.shape != other.shape:
                print("Arrays must have the same shape.")
                return

            if np.any(other == 0):
                print("Division by zero is not allowed.")
                return

            print("Result:")
            print(self.__data / other)

        elif choice == "5":

            if self.__data.size == other.size:
                print("Dot Product:")
                print(np.dot(self.__data,other))
            else:
                print("Arrays must have the same number of elements.")

        elif choice == "6":

            if self.__data.ndim != 2 or other.ndim != 2:
                print("Both arrays must be 2D.")
                return

            if self.__data.shape[1] == other.shape[0]:
                print("Matrix Multiplication:")
                print(np.matmul(self.__data,other))
            else:
                print("Matrix dimensions are not compatible.")

        else:
            print("Invalid choice.")

    # ---------------- SEARCH SORT FILTER ----------------

    def search_sort_filter(self):

        print("\nSearch, Sort, or Filter")
        print("1. Search")
        print("2. Sort")
        print("3. Filter")

        choice = input("Enter choice: ")

        if choice == "1":

            value = float(input("Enter value to search: "))

            result = np.where(self.__data == value)

            if result[0].size > 0:
                print("Value found at:",result)
            else:
                print("Value not found.")

        elif choice == "2":

            order = input("Enter A for ascending or D for descending: ")

            if order.upper() == "A":
                print("Sorted Array:")
                print(np.sort(self.__data))

            elif order.upper() == "D":
                print("Sorted Array:")
                print(np.sort(self.__data)[::-1])

            else:
                print("Invalid choice.")

        elif choice == "3":

            value = float(input("Show values greater than: "))

            print("Filtered Array:")
            print(self.__data[self.__data > value])

        else:
            print("Invalid choice.")

    # ---------------- AGGREGATES AND STATISTICS ----------------

    def aggregates_statistics(self):

        print("\nAggregates and Statistics")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum")
        print("7. Maximum")
        print("8. Percentile")
        print("9. Correlation")

        choice = input("Enter choice: ")

        if choice == "1":
            print("Sum:",np.sum(self.__data))

        elif choice == "2":
            print("Mean:",np.mean(self.__data))

        elif choice == "3":
            print("Median:",np.median(self.__data))

        elif choice == "4":
            print("Standard Deviation:",np.std(self.__data))

        elif choice == "5":
            print("Variance:",np.var(self.__data))

        elif choice == "6":
            print("Minimum:",np.min(self.__data))

        elif choice == "7":
            print("Maximum:",np.max(self.__data))

        elif choice == "8":

            value = float(input("Enter percentile from 0 to 100: "))

            if 0 <= value <= 100:
                print("Percentile:",np.percentile(self.__data,value))
            else:
                print("Percentile must be between 0 and 100.")

        elif choice == "9":

            print("\nEnter second array:")
            other = DataAnalytics.create_array()

            if other is None:
                return

            if self.__data.size == other.size:
                print("Correlation:")
                print(
                    np.corrcoef(
                        self.__data.flatten(),
                        other.flatten()
                    )
                )
            else:
                print("Arrays must have the same number of elements.")

        else:
            print("Invalid choice.")


# =========================================================
# MAIN PROGRAM
# =========================================================

def main():

    print("\nWelcome to the NumPy Analyzer!")
    print("========================================")

    data = DataAnalytics.create_array()

    if data is None:
        return

    analyzer = DataAnalytics(data)

    while True:

        print("\nChoose an option:")
        print("1. Create a Numpy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")

        choice = input("Enter your choice: ")

        # Create array
        if choice == "1":

            new_data = DataAnalytics.create_array()

            if new_data is not None:
                analyzer = DataAnalytics(new_data)
                print("\nNew array created successfully.")
                print(analyzer._DataAnalytics__data)

        # Mathematical operations
        elif choice == "2":
            analyzer.mathematical_operations()

        # Combine or split
        elif choice == "3":

            print("\nCombine or Split Arrays")
            print("1. Combine Arrays")
            print("2. Split Array")
            print("3. Indexing")
            print("4. Slicing")

            option = input("Enter choice: ")

            if option == "1":

                print("\nEnter second array:")
                other = DataAnalytics.create_array()

                if other is not None:
                    analyzer.combine(other)

            elif option == "2":
                analyzer.split()

            elif option == "3":
                analyzer.indexing()

            elif option == "4":
                analyzer.slicing()

            else:
                print("Invalid choice.")

        # Search, sort, filter
        elif choice == "4":
            analyzer.search_sort_filter()

        # Aggregates and statistics
        elif choice == "5":
            analyzer.aggregates_statistics()

        # Exit
        elif choice == "6":
            print("\nThank you for using NumPy Analyzer!")
            break

        else:
            print("Invalid choice. Please try again.")


main()
```
